```python
# Linear Regression
# Predicts a continuoous output variable from input feature
from sklearn.linear_model import LinearRegression
```

```python
area = [[1000], [1400], [1600], [4400]]
price = [5, 7, 8, 22]
```

```python
model = LinearRegression()
model.fit(area, price)
```

```python
prediction = model.predict([[3200]])
print(f"Predicted price: {prediction}")
```
