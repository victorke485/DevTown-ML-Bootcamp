```python
# Decision Trees
# A tree-shaped model that makes decisions through a series of questions

from sklearn.tree import DecisionTreeClassifier
```

```python
# Loan approval prediction
salary = [[20], [30], [40], [60]]
loan_status = [0, 0, 1, 1]
model = DecisionTreeClassifier()
model.fit(salary, loan_status)
prediction = model.predict([[40]])
print(prediction)
```
