```python
# Linear Regression
# Predicts a continuoous output variable from input feature
from sklearn.linear_model import LinearRegression
```

```python
# Predict the no. of hours the battery will last based on the battery capacity
battery = [[2000], [3000], [7000], [9000]]
backup = [8, 12, 16, 20]
model = LinearRegression()
model.fit(battery, backup)
prediction = model.predict([[8000]])
print(prediction)
```
